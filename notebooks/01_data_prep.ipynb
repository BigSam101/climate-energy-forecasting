PREPARE AND SET UP THE ENVIRONMENT:
This script prepares and sets up the environment. 
It includes steps to create a virtual environment, install dependencies, and configure necessary settings.

```python
import sys
sys.path.append("../src")

import pandas as pd

from data_utils import (
    create_log_features,
    create_lag_features,
    create_temperature_features,
    create_calendar_dummies,
    split_data
)

# Load data
df = pd.read_csv("../data/nordic_energy_climate_df.csv")

countries = {
    "Norway": "no",
    "Sweden": "se",
    "Finland": "fi"
}

# Feature engineering
df = create_log_features(df, countries)
df = create_lag_features(df, countries)
df = create_temperature_features(df)
df = create_calendar_dummies(df)

df = df.dropna().reset_index(drop=True)

# Time index
df['Load Date'] = pd.to_datetime(df['Load Date'])
df = df.set_index('Load Date').sort_index()

# Split
train_data, val_data, development_data, forecast_data = split_data(
    df,
    train_start_year=2010,
    forecast_year=2025
)
```

```python
output_path = "../data/processed_data.csv"
df.to_csv(output_path)

print(f"Saved processed data to {output_path}")
```
